# Problem 1  
The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves performance.

### Step 1: Choose the correct statistical test
    We have one sample of 50 students.
    We compare their sample average (68) against a historical benchmark (65).
    We know the sample standard deviation (10), but the population standard deviation is not given.
    We want to check whether the sample average is greater than 65.
    We will use a one-sample, right-tailed t-test.

### Step 2: Formulate the hypothesis
   Null hypothesis (H0): The true average score is 65.
   
   Alternative hypothesis (H1): The true average score is greater than 65.

```python
from scipy import stats
import numpy as np
```

```python
# Given information
historical_mean = 65
sample_mean = 68
sample_std = 10
n = 50
alpha = 0.05 #(Level of significance)
```

```python
# Step 1: Calculate t-statistic value
standard_error = sample_std / np.sqrt(n)
t_statistic = (sample_mean - historical_mean) / standard_error
t_statistic
```

```python
# Step 2: Calculate right-tailed p-value
p_value = stats.t.sf(t_statistic, n-1)
p_value
```

```python
# Step 3: Calculate critical t-value
critical_t = stats.t.ppf(1 - alpha, n-1)
critical_t
```

```python
# Display results
print(f"Standard error: {standard_error:.4f}")
print(f"T-statistic: {t_statistic:.4f}")
print(f"Degrees of freedom: {n-1}")
print(f"Critical t-value: {critical_t:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: Significant evidence of improved scores.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: Insufficient evidence of improved scores.")
```

### Business conclusion:
At the 5% significance level, there is sufficient statistical evidence to conclude that the average exam score of students trained under the new course is greater than the historical average of 65.

# Problem 3
  A company claims that 60% of its users are satisfied with its product. In a survey of 200 users, 134 users report satisfaction. At the 5% level of significance, test whether customer satisfaction is higher than claimed.

### Step 1: Choose the correct statistical test
    We are testing a percentage of satisfied users.
    Z-test: We have a large sample of 200 users, with enough expected satisfied and unsatisfied users under the null hypothesis.
    We want to know whether satisfaction is higher than 60%.
    We use a one-proportion, right-tailed z-test.

### Step 2: Formulate the hypothesis
   Null hypothesis (H0): Customer satisfaction is 60%.
   
   Alternative hypothesis (H1): Customer satisfaction is greater than 60%.

```python
# Given data
claimed_proportion = 0.60
total_users = 200
satisfied_users = 134
alpha = 0.05
```

```python
# Step 1: Calculate sample proportion
sample_proportion = satisfied_users / total_users
sample_proportion
```

```python
# Step 2: Calculate z-statistic
z_statistic = (sample_proportion - claimed_proportion) / np.sqrt(claimed_proportion * (1 - claimed_proportion) / total_users)
z_statistic 
```

```python
# Step 3: Calculate right-tailed P-value
p_value = stats.norm.sf(z_statistic)
p_value
```

```python
# Step 4: Calculate critical z-value
critical_z = stats.norm.ppf(1 - alpha)
critical_z
```

```python
# Display results
print("Sample proportion:", round(sample_proportion, 4))
print("Sample satisfaction rate:", f"{sample_proportion:.1%}")
print("Standard error:", round(standard_error, 4))
print("Calculated z-value:", round(z_statistic, 4))
print("P-value:", round(p_value, 4))
print("Critical z-value:", round(critical_z, 4))
```

```python
if p_value < alpha:
    print("Decision: Reject the Null Hypothesis")
    print("Conclusion: Customer satisfaction is significantly higher than 60%.")
else:
    print("Decision: Fail to reject the Null Hypothesis")
    print("Conclusion: Insufficient evidence that satisfaction exceeds 60%.")
```

### Business conclusion
At the 5% significance level, there is sufficient statistical evidence that customer satisfaction is higher than the company's claimed 60%.

# Problem 5
  Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods.

### Step 1: Choose the correct statistical test
    We have two separate groups of employees.
    We are comparing their average scores.
    The population standard deviations are unknown; we have sample standard deviations instead.
    We are checking for any difference, not only whether one method is better.
    We use an Welch's independent two-sample t-test.
    Welch's t-test is a good default because it does not require the two population variances to be equal.

### Step 2: Formulate the hypothesis
   Null hypothesis (H0): There is no difference in the population average scores between the two methods.
   
   Alternative hypothesis (H1): There is a difference in the population average scores between the two methods.

```python
# Given values
mean_A = 75
std_A = 12
n_A = 30

mean_B = 70
std_B = 10
n_B = 28

alpha = 0.05
```

```python
# Step 1 : Perform independent two-sample Welch's t-test
t_value, p_value = stats.ttest_ind_from_stats(mean1=mean_A, std1=std_A, nobs1=n_A, mean2=mean_B, std2=std_B, nobs2 = 28, equal_var=False)
print(t_value,",", p_value)
```

```python
# Step 2 : Calculate Welch's degrees of freedom
var_A = std_A**2 / n_A
var_B = std_B**2 / n_B
print(var_A,",", var_B)
```

```python
df = (var_A + var_B)**2 / (var_A**2 / (n_A - 1)+ var_B**2 / (n_B - 1))
df
```

```python
# Step 3 :Critical value for a two-tailed test
critical_t = stats.t.ppf(1 - alpha / 2, df)
critical_t
```

```python
print("Mean difference:", mean_A - mean_B)
print("Calculated t-value:", round(t_value, 4))
print("Degrees of freedom:", round(df, 2))
print("P-value:", round(p_value, 4))
print("Critical t-value:", round(critical_t, 4))
```

```python
if p_value < alpha:
    print("Decision: Reject the Null Hypothesis")
    print("Conclusion: There is a significant difference between methods.")
else:
    print("Decision: Fail to reject the Null Hypothesis")
    print("Conclusion: Insufficient evidence of a difference between methods.")
```

### Business conclusion
 Method A achieved a sample average score 5 points higher than Method B, there is insufficient statistical evidence at the 5% significance level to conclude that the two teaching methods differ in their population average scores.

# Problem 7
 Two versions of a website are tested. Out of 500 users for Version A, 320 users click the signup button. Out of 480 users for Version B, 290 users click the signup button. At the 5% level of significance, test whether the conversion rates differ.

### Step 1: Choose the correct statistical test
    We are comparing two independent groups: users who visited Version A and users who visited Version B.
    We want to compare two proportions: the conversion rates.
    In the question asks whether the rates differ, not specifically whether A is better than B. Therefore, we use a Two-proportion two-tailed Z-test

### Step 2: Formulate the hypothesis
   Null hypothesis (H0): There is no difference between the true conversion rates of Version A and Version B.
   
   Alternative hypothesis (H1): There is a difference between the true conversion rates of Version A and Version B.

```python
from statsmodels.stats.proportion import proportions_ztest
from scipy.stats import norm
```

```python
# Given values
clicks = [320, 290]
users = [500, 480]

alpha = 0.05
```

```python
# Step 1: Calculate conversion rates
rate_a = (clicks[0] / users[0])* 100
rate_b = (clicks[1] / users[1])* 100

print("Version A conversion rate:", round(rate_a, 2), "%")
print("Version B conversion rate:", round(rate_b, 2), "%")
```

```python
# Step 2: Two-proportion Z-test
z_stat, p_value = proportions_ztest(count=clicks, nobs=users, alternative="two-sided")
print(z_stat, p_value)
```

```python
# Step 3: Critical Z-value for a two-tailed test
critical_z = norm.ppf(1 - alpha / 2)
float(critical_z)
```

```python
print(f"Version A conversion rate: {round(rate_a, 2)}")
print(f"Version B conversion rate: {round(rate_b, 2)}")
print(f"Z-statistic: {z_stat:.4f}")
print(f"Critical Z-value: +/-{critical_z:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
# Decision using p-value
if p_value < alpha:
    print("Decision: Reject the Null Hypothesis")
    print("Conclusion: There is sufficient statistical evidence at the 5% significance level to conclude that the mean delivery times differ significantly.")
else:
    print("Decision: Fail to reject the Null Hypothesis")
    print("Conclusion: There is insufficient statistical evidence at the 5% significance level to conclude that the mean delivery times differ.")
```

### Business conclusion
Version A achieved a sample conversion rate 3.58 percentage points higher than Version B. However, there is insufficient statistical evidence at the 5% significance level to conclude that the two website versions differ in their population conversion rates.

# Problem 9
A company wants to compare the average productivity of employees trained using three different training programs. Program A has 25 employees with a mean productivity score of 80 and a standard deviation of 8. Program B has 30 employees with a mean of 75 and a standard deviation of 7. Program C has 28 employees with a mean of 82 and a standard deviation of 6. At the 5% level of significance, test whether there is a difference in mean productivity across the programs

### Step 1: Choose the correct statistical test
    We are comparing three independent groups.
    Productivity scores are numerical.
    We want to test whether at least one population mean differs from the others.
    We use One-Way ANOVA (Analysis of Variance).
    One-way ANOVA compares two kinds of variation:
        - Between-group variation: How far the program means are from the overall mean.
        - Within-group variation: How much individual employee scores vary within each program.

### Step 2: Formulate the hypothesis
   Null hypothesis (H0): All three population means are equal.
   
   Alternative hypothesis (H1): At least one population mean differs.

```python
from scipy.stats import f
```

```python
# Given data
n = [25, 30, 28]
means = [80, 75, 82]
stds = [8, 7, 6]

alpha = 0.05

# Total number of employees and groups
N = sum(n)
k = len(n)
```

```python
# The grand mean is the overall average productivity score across all employees.
# Step 1: Weighted grand mean
grand_mean = sum(n[i] * means[i] for i in range(k)) / N
grand_mean
```

```python
# Step 2:  Between-group sum of squares
ss_between = sum(n[i] * (means[i] - grand_mean) ** 2 
                 for i in range(k))
ss_between
```

```python
# Step 3:  Within-group sum of squares
ss_within = sum((n[i] - 1) * stds[i] ** 2
             for i in range(k))
ss_within
```

```python
# Degrees of freedom
df_between = k - 1
df_within = N - k

# Mean squares
ms_between = ss_between / df_between
ms_within = ss_within / df_within
```

```python
# F-statistic
f_stat = ms_between / ms_within

# Critical F-value and p-value
critical_f = f.ppf(1 - alpha, df_between, df_within)
p_value = f.sf(f_stat, df_between, df_within)
```

```python
print(f"Grand mean: {grand_mean:.4f}")
print(f"Between-group SS: {ss_between:.4f}")
print(f"Within-group SS: {ss_within:.4f}")
print(f"F-statistic: {f_stat:.4f}")
print(f"Critical F-value: {critical_f:.4f}")
print(f"P-value: {p_value:.6f}")
```

```python
if p_value < alpha:
    print("Decision: Reject the Null Hypothesis")
    print("Conclusion: There is sufficient statistical evidence that mean productivity differs across the programs.")
else:
    print("Decision: Fail to reject the Null Hypothesis")
    print("Conclusion: There is insufficient statistical evidence that mean productivity differs across the programs.")
```

### Business conclusion
At the 5% significance level, there is sufficient statistical evidence to conclude that the population mean productivity scores are not equal across the three training programs. The result suggests that the choice of training program may be associated with differences in employee productivity.

# Problem 10
A company wants to compare the average productivity of employees trained using three different training programs. Program A has 25 employees with a mean productivity score of 80 and a standard deviation of 8. Program B has 30 employees with a mean of 75 and a standard deviation of 7. Program C has 28 employees with a mean of 82 and a standard deviation of 6. At the 5% level of significance, test whether there is a difference in mean productivity across the programs

### Step 1: Choose the correct statistical test
    We have two categorical variables: gender and product preference.
    We want to determine whether there is an association between the two variables.
    The question is, Does product preference depend on gender, or are they independent?
    We use Chi-Square Test of Independence.

### Step 2: Formulate the hypothesis
   Null hypothesis (H0): Gender and product preference are independent. There is no association between them.
   
   Alternative hypothesis (H1): Gender and product preference are not independent. There is an association between them.

```python
from math import erfc, sqrt
```

```python
# Given Data
# Observed frequencies
observed = [[60, 40],
            [50, 50]]
```

```python
# Step 1 : Calculate row totals, column totals and grand total
row_totals = [sum(row) for row in observed]
row_totals
```

```python
column_totals = [sum(row[j] for row in observed) for j in range(2)]
column_totals
```

```python
grand_total = sum(row_totals)
grand_total
```

```python
# Step 2 : Calculate expected frequencies
expected = [[row_totals[i] * column_totals[j] / grand_total for j in range(2)] for i in range(2)]
expected
```

```python
# Step 3 : Calculate Chi-Square statistic
chi_square = sum((observed[i][j] - expected[i][j]) ** 2 / expected[i][j] for i in range(2) for j in range(2))
chi_square
```

```python
# Degrees of freedom
df = (len(observed) - 1) * (len(observed[0]) - 1)
df
```

```python
# For Chi-Square with 1 degree of freedom
p_value = erfc(sqrt(chi_square / 2))
p_value
```

```python
print("Expected frequencies:", expected)
print(f"Chi-Square statistic: {chi_square:.4f}")
print(f"Degrees of freedom: {df}")
print(f"P-value: {p_value:.4f}")
```

```python
if p_value < 0.05:
    print("Decision: Reject the Null Hypothesis")
    print("Conclusion: Gender and product preference are associated.")
else:
    print("Decision: Fail to reject the Null Hypothesis")
    print("Conclusion: Insufficient evidence of an association.")
```

### Business conclusion
At the 5% significance level, there is insufficient statistical evidence to conclude that gender and product preference are associated. Although the observed preferences differ between males and females, the difference is not statistically significant. Therefore, the company cannot conclude from this survey that product preference depends on gender.

We have two categorical variables: smoking status and disease occurrence.

- Smoking status: Smokers and non-smokers.
- Disease occurrence: Disease and no disease.
- We want to check whether these two variables are independent.
- We will use a **Chi-Square Test of Independence**.

### Step 2: Formulate the hypotheses

**Null hypothesis (H₀):** Smoking status and disease occurrence are independent.

**Alternative hypothesis (H₁):** Smoking status and disease occurrence are associated (not independent).

Significance level: α = 0.05 (5%).

```python
# Given Data
observed = np.array([
    [70, 30],  # Smokers: disease, no disease
    [40, 60]   # Non-smokers: disease, no disease
])
alpha = 0.05
```

```python
# Step 1: Calculate chi-square statistic and p-value
chi2_statistic, p_value, dof, expected = (stats.chi2_contingency(observed, correction=False))
chi2_statistic, p_value, dof, expected
```

```python
# Step 2: Calculate critical chi-square value
critical_chi2 = stats.chi2.ppf(1 - alpha, dof)
```

```python
# Display results
print("Expected frequencies:")
print(expected)
print(f"Chi-square statistic: {chi2_statistic:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"Critical chi-square value: {critical_chi2:.4f}")
print(f"P-value: {p_value:.6f}")
```

```python
# Step 3: Make the decision
if p_value < alpha:
    print("Decision: Reject the Null Hypothesis")
    print("Conclusion: Smoking status and disease occurrence are associated.")
else:
    print("Decision: Fail to reject Null Hypothesis")
    print("Conclusion: Insufficient evidence of an association.")
```

### Business conclusion

# Problem 12
A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
●	Face 1: 18
●	Face 2: 22
●	Face 3: 20
●	Face 4: 25
●	Face 5: 17
●	Face 6: 18
At the 5% level of significance, test whether the die is fair


### Step 1: Choose the correct statistical test
- We have one categorical variable: dice outcomes (1, 2, 3, 4, 5, 6).
- The die is rolled 120 times.
- We have observed frequencies for all six faces.
- If the die is fair, each face should have an equal probability of occurring (1/6).
- We want to check whether the observed frequencies follow the expected uniform distribution.
- We will use a Chi-Square Goodness-of-Fit Test.

### Step 2: Formulate the hypotheses
Null hypothesis (H₀): The die is fair. All six faces have an equal probability of occurring (1/6).

Alternative hypothesis (H₁): The die is not fair. At least one face has a different probability of occurring.

Significance level: (5%).

```python
# Given information
observed = np.array([18, 22, 20, 25, 17, 18])

total_rolls = 120
number_of_faces = 6
alpha = 0.05
```

```python
# Step 1: Calculate expected frequencies
expected = np.full(number_of_faces, total_rolls / number_of_faces)
expected
```

```python
# Step 2: Calculate chi-square statistic and p-value
chi2_statistic, p_value = stats.chisquare(f_obs=observed, f_exp=expected)
chi2_statistic, p_value
```

```python
# Step 3: Calculate degrees of freedom
dof = number_of_faces - 1
dof
```

```python
# Step 4: Calculate critical chi-square value
critical_chi2 = stats.chi2.ppf(1 - alpha, dof)

# Display results
print("Expected frequencies:", expected)
print(f"Chi-square statistic: {chi2_statistic:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"Critical chi-square value: {critical_chi2:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
# Step 5: Make the decision
if p_value < alpha:
    print("Decision: Reject the Null Hypothesis")
    print("Conclusion: The die is not fair.")
else:
    print("Decision: Fail to reject Null Hypothesis")
    print("Conclusion: Insufficient evidence that the die is unfair.")
```

At the 5% significance level, there is insufficient statistical evidence to conclude that the die is unfair.
